```python
def graficar_pinguinos(ind):

    # Eliminamos datos faltantes y reseteamos los índices, para no tener problemas al graficar
    penguins = sns.load_dataset("penguins").dropna().reset_index(drop=True)
    
    # Normalizamos las variables "bill_length_mm" y "bill_depth_mm" por MinMax
    penguins[["bill_length_mm", "bill_depth_mm"]] = MinMaxScaler().fit_transform(penguins[["bill_length_mm", "bill_depth_mm"]])

    neighbors = NearestNeighbors(n_neighbors=10)  # Esta función nos devuelve los más cercanos incluyendo a si mismo, por eso tomamos 10.
    neighbors_fit = neighbors.fit(datos)
    
    # Ordenamos de menor a mayor las distancias y graficamos
    distances, indices = neighbors_fit.kneighbors(datos.iloc[[ind]])

    # Graficamos
    plt = (
        so.Plot(data = penguins.iloc[indices.flatten()], x = "bill_length_mm", y = "bill_depth_mm",  text = np.arange(10).astype(str))
        .add(so.Text(valign = "bottom"))
        .add(so.Dot(), color = "species")
        .add(so.Dot(color = "red"), data = penguins.iloc[[ind]], x = "bill_length_mm", y = "bill_depth_mm")
    )
    return(plt)
```

```python
plt = graficar_pinguinos(15)
display(plt)
```

```python
plt = graficar_pinguinos(151)
display(plt)
```

```python
plt = graficar_pinguinos(313)
display(plt)
```

```python

for K in range(1,30):
    
    neighbors = NearestNeighbors(n_neighbors= K)  # Esta función nos devuelve los más cercanos incluyendo a si mismo, por eso tomamos 10.
    
    correctos = 0
    for ind in range(len(penguins)):
        distances, indices = neighbors.fit(datos.drop([ind])).kneighbors(datos.iloc[[ind]])
        if(penguins["species"][ind] == mode(penguins["species"][indices.flatten()])):
            correctos += 1
    
    print(K, correctos)
```
